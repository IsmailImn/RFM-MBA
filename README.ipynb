# Customer Segmentation and Market Basket Analysis  
*(Online Retail Dataset – United Kingdom)*

## 📌 Project Overview
Proyek ini bertujuan untuk memahami perilaku pelanggan berdasarkan data transaksi penjualan melalui pendekatan **Customer Segmentation (RFM Analysis)** dan **Market Basket Analysis (Apriori Algorithm)**.

Analisis difokuskan pada transaksi di wilayah United Kingdom untuk mengidentifikasi pelanggan bernilai tinggi serta pola pembelian produk yang sering terjadi bersamaan.

---

## 🎯 Project Objectives
- Mengelompokkan pelanggan berdasarkan perilaku pembelian menggunakan metode RFM
- Mengidentifikasi pola asosiasi antar produk menggunakan Market Basket Analysis
- Memberikan insight dan rekomendasi bisnis berbasis data untuk meningkatkan penjualan

---

## 📂 Dataset
- **Dataset**: Online Retail Dataset
- **Periode Data**: 2010 – 2011
- **Atribut Utama**:
  - InvoiceNo
  - InvoiceDate
  - CustomerID
  - Quantity
  - UnitPrice
  - Country
  - Description (Product)

---

## 🧹 Data Cleaning
Tahap pembersihan data dilakukan untuk memastikan analisis yang akurat, meliputi:
- Penghapusan missing values dan data duplikat
- Penghapusan transaksi pembatalan (InvoiceNo mengandung huruf "C")
- Penghapusan transaksi dengan Quantity ≤ 0 dan UnitPrice = 0

Dataset hasil pembersihan digunakan sebagai dasar analisis RFM dan Market Basket Analysis.

---

## 📊 Methodology
### 1. RFM Analysis
- **Recency**: Jarak waktu sejak transaksi terakhir
- **Frequency**: Jumlah transaksi pelanggan
- **Monetary**: Total nilai belanja pelanggan  
(Log transformation diterapkan untuk mengatasi distribusi ekstrem)

### 2. Market Basket Analysis
- Algoritma: Apriori
- Threshold utama:
  - Support ≥ 3%
  - Confidence ≥ 60%
  - Lift > 1
- Fokus pada association rules yang kuat dan relevan secara bisnis

---

## 🔍 Key Insights
- Pelanggan dapat diklasifikasikan ke dalam segmen Loyal, Potential, New Customer, dan At Risk
- Ditemukan 12 association rules berkualitas tinggi yang menunjukkan hubungan pembelian produk yang signifikan
- Beberapa produk memiliki keterkaitan kuat dan berpotensi untuk strategi cross-selling dan bundling

---

## 💡 Business Recommendations
- Menerapkan rekomendasi produk berbasis association rules
- Membuat paket bundling produk dengan tingkat konversi tinggi
- Mengoptimalkan strategi pemasaran berdasarkan segmen pelanggan RFM
- Meningkatkan nilai transaksi pelanggan Potential melalui promosi terarah
